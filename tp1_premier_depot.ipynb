# TP 1 : Mon premier dépôt de projet

**Cours :** Data Science

**Objectif :** Découvrir les notebooks Python, effectuer un calcul et représenter des données graphiquement.

```python
# Calcul de la moyenne de trois notes

notes = [12, 15, 18]

moyenne = sum(notes) / len(notes)

print("Les notes sont :", notes)
print("La moyenne est :", moyenne)
```

```python
import matplotlib.pyplot as plt

notes = [12, 15, 18]
etudiants = ["Étudiant 1", "Étudiant 2", "Étudiant 3"]

plt.bar(etudiants, notes)

plt.title("Notes des étudiants")
plt.xlabel("Étudiants")
plt.ylabel("Notes sur 20")

plt.show()
```
